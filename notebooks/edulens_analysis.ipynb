```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import classification_report, confusion_matrix
```

```python
np.random.seed(42)
n = 200

moyenne = np.random.normal(12, 4, n).clip(0, 20)
absences = np.random.randint(0, 15, n)
absences_non_justifiees = np.random.randint(0, absences + 1, n)

df = pd.DataFrame({
    'moyenne': moyenne,
    'absences': absences,
    'absences_non_justifiees': absences_non_justifiees
})

df.head()
```

```python
df['a_risque'] = ((df['moyenne'] < 10) | (df['absences_non_justifiees'] > 5)).astype(int)
df['a_risque'].value_counts()
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
sns.histplot(df['moyenne'], bins=20, color='steelblue')
plt.title('Distribution des moyennes')

plt.subplot(1, 3, 2)
sns.histplot(df['absences'], bins=15, color='orange')
plt.title('Distribution des absences')

plt.subplot(1, 3, 3)
sns.countplot(x='a_risque', data=df, palette=['green', 'red'])
plt.title('Étudiants à risque')

plt.tight_layout()
plt.show()
```
